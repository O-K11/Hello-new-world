## NumPy neden önemli?

NumPy, Python’da **sayısal hesap** ve **çok boyutlu veri** (vektör/matris) işlemleri için temel kütüphanedir. Özellikle tamamen ya da ağırlıklı olarak **numerik** veriyle çalışırken hız ve verimlilik sağlar.

### 1) Hız ve performans (neden “hızlı”?)
- NumPy’ın temel veri yapısı `ndarray` **tek bir dtype** (homogeneous) taşır: `int32`, `float64` gibi.
- Tek tip veri sayesinde:
  - Bellekte daha **kompakt** tutulur (daha az “ek bilgi/metadata” taşır).
  - Vektörize işlemler ve lineer cebir operasyonları genelde daha **hızlı** çalışır.
- Bu yüzden birçok kütüphane (ör. **scikit-learn**) veri girişinde NumPy array’lerini yaygın olarak kullanır; **lineer cebir** ve model matrisi mantığına uyumu yüksektir.

### 2) NumPy vs Pandas: Ne zaman hangisi?
- **Matematik / model matrisi / büyük sayısal hesap** → **NumPy**
- **Veri temizleme, birleştirme, gruplayıp raporlama** → **Pandas**

### 3) Pandas’ta olup NumPy’da olmayan “tablo özellikleri”
Pandas daha “Excel gibi” çalışır:
- **Kolon isimleri** ve **satır indeksleri** (örn. `Yaş`, `Cinsiyet` gibi sütunlar; `Ahmet`, `Mehmet` gibi satır etiketleri)
- **Join/Merge** (tablo birleştirme)
- **GroupBy** (gruplayıp özetleme)
- **Time series resampling** (zaman serisini haftalık/aylık toplama vb.)
- **Eksik veri/etiket hizalama** (index üzerinden otomatik hizalama)
- **Heterojen tablo**: DataFrame’de her sütun farklı dtype olabilir

NumPy ise daha “ham” ve daha düşük seviyelidir:
- Kolon/satır etiketleri yerine **pozisyon** (0,1,2…) ile çalışır.
- Çok hızlıdır ama “tablo” ergonomisi Pandas kadar güçlü değildir.

### 4) NumPy neden “ara basamak” gibi?
Listelerden Pandas’a giden yolda NumPy kritik bir geçiştir:
- **Vektör ve matris** mantığını öğretir.
- Çok boyutlu veri (`2D`, `3D`, …) üzerinde dilimleme, broadcast, toplama/çarpma gibi işlemler için çok işlevseldir.

### 5) Görünüm: “ham” ve daha teknik
- NumPy dizileri çoğu zaman iç içe geçmiş, düzenli (simetrik) listeler gibi görünür.
- Bu yüzden ilk bakışta Pandas’a göre daha “korkutucu” gelebilir; ama sayısal hesap tarafında çok güçlüdür.

12- Belirli bir dizi (`dizi`) ve istediğiniz bir adım (veya pencere boyutu) için bir `time_resolved` fonksiyonu yazın.

Fonksiyon şu üç parametreyi alacak:  
`time_resolved(dizi, pencere_genişliği, kayma)`

Örnek olarak,  
`dizi = np.array([1, 2, 6, 4, 5, 4, 2, 7, 8, 9])` için  
`time_resolved(dizi, 3, 4)` çağrıldığında, fonksiyonun çıktısı `[4, 5, 4]` olmalıdır.

````python
def time_resolved(dizi,pencere_genişliği,kayma):
    alt_dizi=np.array([])
    for i in range(kayma-1,kayma+pencere_genişliği-1):
        alt_dizi=np.append(alt_dizi,dizi[i])
    return alt_dizi
````

````python
dizi = np.array([1, 2, 6, 4, 5, 4, 2, 7, 8, 9])
time_resolved(dizi, 3, 4)
````

Şimdi önemli sözdizimlerini öğrenmek için bazı basit grafikler oluşturalım.



### Matplotlib.pyplot — Mini Cheat Sheet (tek sayfa)

**Kurulum**
- `import matplotlib.pyplot as plt` → pyplot’u kullanmaya başla

**Çizim (grafik türü seç)**
- `plt.plot(x, y)` → çizgi (trend/zaman serisi)
- `plt.bar(x, h)` → bar/sütun (kategori karşılaştırma)
- `plt.hist(data, bins=...)` → histogram (dağılım)
- `plt.scatter(x, y)` → scatter (ilişki)

**Anlatı / Okunabilirlik**
- `plt.title("...")` → başlık
- `plt.xlabel("...")` / `plt.ylabel("...")` → eksen isimleri (birim ekle)
- `plt.legend()` → seri açıklaması (önce `label="..."` vermiş ol)
- `plt.grid(True)` → ızgara (opsiyonel)

**Düzen**
- `plt.tight_layout()` → taşmaları düzelt, boşlukları ayarla

**Çıktı**
- `plt.savefig("fig.png", dpi=300)` → dosyaya kaydet (rapor/tez için)
- `plt.show()` → ekranda göster (genelde en sonda)

**Tipik akış  
**Draw → Label → Layout → Save → Show** ✅


````python
import matplotlib.pyplot as plt

X = [1,2,3,4]
Y = [10,22,30,40]

plt.plot(X, Y, 'b--1')

plt.xlabel('X label')

plt.ylabel('Y label')

plt.title('Bu bir başlıktır')

plt.show()
````

2- 0 ile 1000 arasında rastgele değerler oluşturun ve bu değerleri grafik üzerinde gösterin.  
Grafikte, 500’den küçük değerler için mavi, 500’den büyük değerler için kırmızı renk kullanın.

````python
import numpy as np

data=np.random.randint(0,1000,50)
x = np.arange(50)

data_kırmızı=np.array([])
data_kırmızı_indeks=np.array([])

for i in range(len(data)):
    if data[i]>500:
        data_kırmızı = np.append(data_kırmızı, data[i])
        data_kırmızı_indeks = np.append(data_kırmızı_indeks, i)

data_mavi=np.array([])
data_mavi_indeks=np.array([])

for i in range(len(data)):
    if data[i]<=500:
        data_mavi = np.append(data_mavi, data[i])
        data_mavi_indeks = np.append(data_mavi_indeks, i)

plt.plot(x,data,"k--", alpha=0.5)
plt.plot(data_mavi_indeks,data_mavi,"b-*")
plt.plot(data_kırmızı_indeks,data_kırmızı,"r-^")
plt.show()
````

3- Numpy alıştırmalarında yazdığınız _time_resolved analysis_ fonksiyonunu, yukarıda yer alan alıştırmadaki veriye uygulayın ve sonucu yeni bir grafikte gösterin.

````python
def time_resolved(dizi,pencere_genişliği,kayma):
    alt_dizi=np.array([])
    for i in range(kayma-1,kayma+pencere_genişliği-1):
        alt_dizi=np.append(alt_dizi,dizi[i])
    return alt_dizi
````

````python
import matplotlib.pyplot as plt

rand_değerler=np.random.randint(0,1000,50)
print(rand_değerler)
sayılar=np.arange(1,51)
print(sayılar)
plt.plot(sayılar, rand_değerler, 'r')

pencere_genişliği = int(input())
kayma = int(input())

alt_dizi = time_resolved(rand_değerler,pencere_genişliği,kayma)
sayılar=np.arange(kayma,pencere_genişliği+kayma)

print(alt_dizi)
plt.plot(sayılar,alt_dizi,'b')
plt.show
````
